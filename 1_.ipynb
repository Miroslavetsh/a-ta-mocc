```python
import numpy as np
import sympy as sp
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
```

```python
x1, x2 = sp.symbols('x1 x2', real=True)
f = 2*x1**2 + 3*x2**2 - 2*x1*x2 - 4*x1 - 6*x2 + 10
```

```python
grad = [sp.diff(f, v) for v in (x1, x2)]
H = sp.hessian(f, (x1, x2))
print("grad f =", grad)
print("H =", H.tolist())
```

```python
crit = sp.solve(grad, [x1, x2], dict=True)[0]
print("стаціонарна точка:", crit)
print("f* =", sp.simplify(f.subs(crit)))
```

```python
minors = [H[:k, :k].det() for k in (1, 2)]
print("кутові мінори:", minors)
print("власні числа:", [sp.nsimplify(v) for v in H.eigenvals()])
print("опукла (мінор > 0):", all(m > 0 for m in minors))
```

```python
fn = sp.lambdify((x1, x2), f, 'numpy')
X, Y = np.meshgrid(np.linspace(-1, 4, 300), np.linspace(-1, 4, 300))
Z = fn(X, Y)
```

```python
fig, ax = plt.subplots(figsize=(6, 5))
cs = ax.contour(X, Y, Z, levels=20)
ax.clabel(cs, inline=True, fontsize=7)
ax.plot(float(crit[x1]), float(crit[x2]), 'r*', ms=14, label='точка мінімуму')
ax.set_xlabel('$x_1$'); ax.set_ylabel('$x_2$'); ax.legend(); ax.grid(alpha=.3)
ax.set_title('Лінії рівня цільової функції')
fig.tight_layout(); fig.savefig('contour.png', dpi=150)
```
